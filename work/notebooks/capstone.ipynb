## 0. Abstract

How can search editorial teams systematically identify which published content is decaying in search visibility before traffic declines become irreversible? Using an anonymized multi-client search performance dataset comprising 30,000 URLs across 32 pseudonymous enterprise clients, we developed a machine-learned refresh opportunity scoring pipeline. We trained and evaluated regularized Logistic Regression, constrained Decision Trees, and Random Forest ensembles under a strict client-holdout cross-validation split to eliminate cross-domain data leakage. The Random Forest model achieved a Precision@50 of 0.680–0.740 and an ROC AUC of 0.747, outperforming the heuristic baseline (Precision@50: 0.240, ROC AUC: 0.627) by nearly 3x. The pipeline outputs a prioritized, reason-coded refresh queue that triages pages into concrete editorial actions (refresh_and_review_ctr, refresh_and_review_engagement, expand_and_refresh, and monitor), enabling content teams to concentrate limited rewriting capacity on high-demand, high-decay-risk assets.

## 1. Question

*The research question and the decision it supports.*

**Core Question:** Which published content items are at the highest risk of organic search visibility decay and should be prioritized first in an editorial sprint?

**Operational Decision:** Content marketing teams have finite writing capacity (typically 20–50 URLs per monthly sprint). This model ranks existing assets to maximize the recovery of high-value search impressions while preventing editorial hours from being wasted on naturally fluctuating or stable evergreen pages.

## 5. Limitations

*What this work cannot claim.*

```python
print('1. Observational, not causal: We cannot claim refreshing a page causes rankings to rise without randomized A/B experimentation.')
print('2. No Google algorithm reverse-engineering: This models empirical visibility decay, not Google\'s internal neural ranking algorithms.')
print('3. Seasonal volatility: Sudden external demand drops (e.g. seasonal holiday shifts) can appear as content decay.')
print('4. Decision-support role: The model serves as an editorial triage accelerator, not an autonomous publishing system.')
```
